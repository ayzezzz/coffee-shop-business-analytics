```python
df["revenue"] = df["transaction_qty"] * df["unit_price"]
```

```python
df["revenue"].sum()
```

```python
store_revenue = (
    df.groupby("store_location")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

store_revenue
```

```python
category_revenue = (
    df.groupby("product_category")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

category_revenue
```

```python
df["hour"] = pd.to_datetime(
    df["transaction_time"],
    format="%H:%M:%S"
).dt.hour

hourly_sales = (
    df.groupby("hour")["revenue"]
    .sum()
)

hourly_sales
```

```python
store_revenue.plot(kind="bar", figsize=(8,5))

plt.title("Revenue by Store")
plt.xlabel("Store")
plt.ylabel("Revenue")
plt.show()
```

```python
store_revenue = (
    df.groupby("store_location", as_index=False)["revenue"]
      .sum()
      .sort_values(by="revenue", ascending=False)
)

store_revenue
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    store_revenue["store_location"],
    store_revenue["revenue"]
)

plt.title("Revenue by Store")
plt.xlabel("Store")
plt.ylabel("Revenue")

plt.show()
```

```python
category_revenue = (
    df.groupby("product_category", as_index=False)["revenue"]
      .sum()
      .sort_values(by="revenue", ascending=False)
)

category_revenue
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    category_revenue["product_category"],
    category_revenue["revenue"]
)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

```python
top_products = (
    df.groupby("product_detail", as_index=False)["revenue"]
      .sum()
      .sort_values(by="revenue", ascending=False)
      .head(10)
)

top_products
```

```python
plt.figure(figsize=(12,6))

plt.barh(
    top_products["product_detail"],
    top_products["revenue"]
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")

plt.gca().invert_yaxis()

plt.show()
```

```python
df["hour"] = df["transaction_time"].apply(lambda x: x.hour)
df[["transaction_time", "hour"]].head()
```

```python
df[["transaction_time", "hour"]].head()
```

```python
hourly_revenue = (
    df.groupby("hour", as_index=False)["revenue"]
      .sum()
)

hourly_revenue
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    hourly_revenue["hour"],
    hourly_revenue["revenue"],
    marker="o"
)

plt.title("Revenue by Hour")
plt.xlabel("Hour")
plt.ylabel("Revenue")

plt.grid(True)

plt.show()
```

```python
df["day_name"] = df["transaction_date"].dt.day_name()
```

```python
weekday_revenue = (
    df.groupby("day_name", as_index=False)["revenue"]
      .sum()
)
```

```python
days = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

weekday_revenue["day_name"] = pd.Categorical(
    weekday_revenue["day_name"],
    categories=days,
    ordered=True
)

weekday_revenue = weekday_revenue.sort_values("day_name")
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    weekday_revenue["day_name"],
    weekday_revenue["revenue"]
)

plt.title("Revenue by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Revenue")

plt.show()
```

```python
df["month"] = df["transaction_date"].dt.month_name()
```

```python
monthly_revenue = (
    df.groupby("month", as_index=False)["revenue"]
      .sum()
)
```

```python
months = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June"
]

monthly_revenue["month"] = pd.Categorical(
    monthly_revenue["month"],
    categories=months,
    ordered=True
)

monthly_revenue = monthly_revenue.sort_values("month")
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_revenue["month"],
    monthly_revenue["revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.grid()

plt.show()
```
