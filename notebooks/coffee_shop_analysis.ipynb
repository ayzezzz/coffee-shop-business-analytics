```python
df["revenue"] = df["transaction_qty"] * df["unit_price"]
```

```python
df["revenue"].sum()
```

```python
store_revenue = (
    df.groupby("store_location")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

store_revenue
```

```python
category_revenue = (
    df.groupby("product_category")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

category_revenue
```

```python
df["hour"] = pd.to_datetime(
    df["transaction_time"],
    format="%H:%M:%S"
).dt.hour

hourly_sales = (
    df.groupby("hour")["revenue"]
    .sum()
)

hourly_sales
```

```python
store_revenue.plot(kind="bar", figsize=(8,5))

plt.title("Revenue by Store")
plt.xlabel("Store")
plt.ylabel("Revenue")
plt.show()
```

```python
import sqlite3

conn = sqlite3.connect("../coffee_shop.db")
```

```python
df.to_sql(
    "sales",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT *
FROM sales
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    ROUND(SUM(transaction_qty * unit_price),2) AS total_revenue
FROM sales;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    store_location,
    ROUND(SUM(transaction_qty * unit_price),2) AS revenue
FROM sales
GROUP BY store_location
ORDER BY revenue DESC;
"""

pd.read_sql(query, conn)
```
